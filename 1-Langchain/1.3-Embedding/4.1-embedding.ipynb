```python
import os
from dotenv import load_dotenv
load_dotenv()  #load all the environment variables
```

```python
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings=OpenAIEmbeddings(model="text-embedding-3-large")
embeddings
```

```python
text = "I will be billionaire";
query_result = embeddings.embed_query(text)
query_result
```

```python
len(query_result)
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings_1024 = OpenAIEmbeddings(model="text-embedding-3-large",dimensions=1024)
###  Example only
result = embeddings_1024.embed_query("I will be billionaire")
print(result)
len(result)
```

```python
from langchain_community.document_loaders import TextLoader

loader=TextLoader("speech.txt")
docs=loader.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter=RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=20)
final_documents=text_splitter.split_documents(docs)
final_documents[0]
final_documents[1]
```

```python
final_documents
```

```python
## Vector Embedding And Vector StoreDB

from langchain_community.vectorstores import Chroma

from langchain_openai import OpenAIEmbeddings

embeddings_3072 = OpenAIEmbeddings(model="text-embedding-3-large",dimensions=3072)

db=Chroma.from_documents(final_documents,embeddings_3072)
db


```

```python
### Retrieve the results from query vectorstore db
query="To such a task we can dedicate o"
retrieved_results=db.similarity_search(query)
print(retrieved_results)
```
